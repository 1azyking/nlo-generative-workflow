# Phase diagram and stability

Analyze formation energy and energy above the convex hull using VASP entries and Materials Project phase-diagram entries.
Set the data and output directories before running the analysis.


```python
from mp_api.client import MPRester
from pymatgen.io.vasp import Vasprun
from pymatgen.analysis.phase_diagram import PhaseDiagram, PDPlotter
```

```python
vasprun_file = "./vasprun_LBC4O8_mp.xml"
# vasprun_file = "./vasprun_L3BCO5_vk.xml"
vasprun = Vasprun(vasprun_file)
entry = vasprun.get_computed_entry(inc_structure=True)
print(entry.composition)
print(entry.energy)
```

```python
sys_elems = entry.composition.chemical_system.split("-")
sys_elems
```

```python
import os
API_KEY = os.environ["MP_API_KEY"]
# API_KEY = "<REDACTED>"
with MPRester(API_KEY) as mpr:
    sys_entries = mpr.get_entries_in_chemsys(elements=sys_elems)
```

```python
all_entries = sys_entries
all_entries.append(entry)
```

```python
pd = PhaseDiagram(all_entries)
```

```python
pd.get_e_above_hull(entry)
```

```python
pd.get_form_energy_per_atom(entry)
```

```python
pd.get_decomposition(entry.composition)
```

```python
pd.get_hull_energy_per_atom(comp=entry.composition)
```

```python
print([e.composition for e in sys_entries])
```

```python
mp_entry = None
for e in sys_entries:
    if e.composition == entry.composition:
        mp_entry = e
print(type(mp_entry))
```

```python
pd.get_e_above_hull(mp_entry)
```
