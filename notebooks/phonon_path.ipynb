# Phonon path

Generate a high-symmetry path from the primitive crystal cell.
Set the data and output directories before running the analysis.


```python
import seekpath
from ase.io import read
import numpy as np
```

```python
poscar_file = "./data/dfpt/BCSCO-POSCAR-opt-221"

atoms = read(poscar_file)  # Primitive cell, not a supercell.

# Use seekpath to identify the space group and high-symmetry path.
cell = (atoms.cell[:], atoms.get_scaled_positions(), atoms.numbers)

# Method 1: use seekpath.
result = seekpath.get_path(cell)
print("Space group:", result['spacegroup_international'])
print("\nHigh-symmetry point coordinates (fractional coordinates):")
for name, coords in result['point_coords'].items():
    print(f"  {name}: {coords}")
print("\nSuggested path:")
for segment in result['path']:
    print(f"  {segment[0]} -> {segment[1]}")
```
